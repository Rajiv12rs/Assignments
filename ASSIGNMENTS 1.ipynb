```python
# University Tech Fest Participant Management

# Initial list of students
students = ["Aarav", "Isha", "Rohan", "Sneha", "Kabir"]

# 1. Display the third student's name
print("Third student:", students[2])

# 2. Change "Rohan" to "Rohan Patil"
students[2] = "Rohan Patil"

# 3. Add two new students
students.append("Ananya")
students.append("Vihaan")

# 4. Remove "Kabir" from the list
students.remove("Kabir")

# 5. Display the total number of registered students
print("Total registered students:", len(students))

# 6. Check whether "Isha" is registered
if "Isha" in students:
    print("Isha is registered.")
else:
    print("Isha is not registered.")

# 7. Arrange the list in alphabetical order
students.sort()

# 8. Create a backup copy of the final list
backup_students = students.copy()

# AI Hackathon participants
hackathon = ["Meera", "Arjun", "Isha"]

# 9. Combine both participant lists
all_participants = students + hackathon

# 10. Display the final lists
print("\nFinal Tech Fest Student List:")
print(students)

print("\nBackup Student List:")
print(backup_students)

print("\nCombined Participant List:")
print(all_participants)
```

```python
# Initial course tuple
courses = ("Python", "Data Science", "Cloud Computing", "Cyber Security", "AI")

# 1. Display the second course
print("Second course:", courses[1])

# 2. Display the last two courses using slicing
print("Last two courses:", courses[-2:])

# 3. Check whether "Python" is present
print("Is Python present?", "Python" in courses)

# 4. Find the index of "Cloud Computing"
print("Index of Cloud Computing:", courses.index("Cloud Computing"))

# 5. Find the total number of courses
print("Total number of courses:", len(courses))

# 6. Add two new courses by creating a new tuple
courses = courses + ("Machine Learning", "Web Development")

# 7. Drop "Cyber Security"
courses = tuple(course for course in courses if course != "Cyber Security")

# 8. Arrange courses alphabetically
sorted_courses = tuple(sorted(courses))

# 9. Create a backup copy
backup_courses = sorted_courses

# 10. Friend's courses
friend_courses = ("AI", "Blockchain", "Python")

# Combine student's and friend's courses
combined_courses = sorted_courses + friend_courses

# Display final results
print("\nFinal course tuple:", courses)
print("Sorted tuple:", sorted_courses)
print("Backup tuple:", backup_courses)
print("Combined tuple:", combined_courses)
```

```python
# List
students = ["Raj", "Amit", "Riya"]
print("Original list:", students)

# Add student
students.append("Neha")
print("After adding:", students)

# Remove student
students.remove("Amit")
print("After removing:", students)

# Update student
students[0] = "Rahul"
print("After updating:", students)

# Tuple
t = tuple(students)
print("Tuple:", t)

# Tuple to list
students = list(t)
print("Tuple converted to list:", students)

# Dictionary
student = {"name": "Raj", "age": 19}
print("Dictionary:", student)

# Add in dictionary
student["city"] = "Pune"
print("After adding:", student)

# Delete from dictionary
del student["age"]
print("After deleting:", student)
```
